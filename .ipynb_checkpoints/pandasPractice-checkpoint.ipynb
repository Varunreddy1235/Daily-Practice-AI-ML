Your task: Create a new column called "Result" using .apply() and lambda:
- Marks 40 or above → "Pass"
- Marks below 40 → "Fail"

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Anu", "Raj"],
    "Marks": [35, 75, 90]
})
df['Result']=df['Marks'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')
df
```


Create a new column called Bonus using apply() and lambda.
Rules:
- Salary of ₹30,000 or more → bonus is 10% of salary.
- Salary below ₹30,000 → bonus is 5% of salary.

```python
df = pd.DataFrame({    "Name": ["Ravi", "Anu", "Raj"],    "Salary": [20000, 35000, 50000]})
df['bonus']=df['Salary'].apply(lambda x: x*0.1 if x>=30000 else x*0.05)
df
```

Your task: Create a "Total" column by multiplying Price by Quantity for each row.

```python
df = pd.DataFrame({
    "Price": [100, 200, 300],
    "Quantity": [2, 3, 4]
})
df['Total']=df['Price']*df['Quantity']
df
```

Next challenge — combine two conditions
Create a "Category" column using apply() and lambda on the Marks column.
Rules:
- Marks 80 or above → "Excellent"
- Marks 40–79 → "Pass"
- Marks below 40 → "Fail"
Write the line of code that creates df["Category"].

```python
df = pd.DataFrame({
    "Name": ["Ravi", "Anu", "Raj", "Priya"],
    "Marks": [35, 55, 75, 95]
})
df['Category']=df['Marks'].apply(lambda x: 'Excellent' if x>=80 else 'Pass' if x>=40 else 'Fail')
df
```

Next challenge — Pandas map() with a dictionary
Let's switch to a different way to transform data.
Create a new column called "Department_Name" using .map() with this mapping:
- "IT" → "Information Technology"
- "HR" → "Human Resources"
- "Sales" → "Sales Department"

```python
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "Sales"]
})
df['Department_Name']=df['Department'].map({
    "IT" : "Information Technology", "HR" : "Human Resources", "Sales" : "Sales Department"})
df
```

Next challenge — combine map() and fillna()
Suppose your data contains an unknown department:
Use .map() to convert the known departments using the same dictionary, and replace unknown department names with "Other".

Your task: Write the code to create df["Department_Name"] so that "Finance" becomes "Other" instead of NaN.

```python
df = pd.DataFrame({
    "Department": ["IT", "HR", "Finance", "Sales"]
})
df['Department_Name']=df['Department'].map({
    "IT" : "Information Technology", "HR" : "Human Resources",  "Sales" : "Sales Department"
}).fillna('Others')
df
```

Memory trick:
- inner → matching rows only.
- left → all rows from the left DataFrame.
- right → all rows from the right DataFrame.
- outer → all rows from both DataFrames.
