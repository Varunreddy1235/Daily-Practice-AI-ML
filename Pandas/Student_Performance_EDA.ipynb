```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.DataFrame({
    "Student": ["Ravi", "Kumar", "Priya", "Anu", "Rahul",
                "Sneha", "Arjun", "Divya", "Kiran", "Meena"],
    "Hours": [2, 3, 6, 1, 5, 4, 7, 2, 5, 6],
    "Marks": [55, 65, 95, 45, 85, 75, 98, 50, 88, 92],
    "Attendance": [70, 75, 98, 60, 90, 85, 99, 65, 92, 95],
    "Department": ["IT", "CSE", "IT", "ECE", "CSE",
                   "IT", "ECE", "CSE", "IT", "ECE"]})
df.head()
```

```python
#Visualize student marks
sns.barplot(data=df, x="Student", y="Marks")
plt.xticks(rotation=45)
plt.title("Student Marks Analysis")
plt.show()
```

```python
#Analyze study hours and marks
sns.scatterplot(data=df, x="Hours", y="Marks",hue="Department")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
#Compare departments
sns.boxplot(data=df, x="Department", y="Marks")
plt.title("Marks Distribution by Department")
plt.show()
```

```python
#Correlation heatmap
sns.heatmap(
    df[["Hours", "Marks", "Attendance"]].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Student Performance Correlation")
plt.show()
```
