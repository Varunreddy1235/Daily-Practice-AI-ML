```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
```

```python
rf_model.fit(X_train,y_train)
```

```python
rf_pred = rf_model.predict(X_test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error,
    r2_score
)

print("MAE:", mean_absolute_error(y_test, rf_pred))
print("MSE:", mean_squared_error(y_test, rf_pred))
print("RMSE:", root_mean_squared_error(y_test, rf_pred))
print("R2 Score:", r2_score(y_test, rf_pred))
```

```python
comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest'],
    'MAE': [
        mean_absolute_error(y_test, y_pred),
        mean_absolute_error(y_test, rf_pred)
    ],
    'RMSE': [
        root_mean_squared_error(y_test, y_pred),
        root_mean_squared_error(y_test, rf_pred)
    ],
    'R2 Score': [
        r2_score(y_test, y_pred),
        r2_score(y_test, rf_pred)
    ]
})

print(comparison)
```

```python
rf_model_2 = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_model_2.fit(X_train, y_train)

rf_pred_2 = rf_model_2.predict(X_test)

print("R2:", r2_score(y_test, rf_pred_2))
```

```python
rf_model_3 = RandomForestRegressor(
    n_estimators=200,
    max_depth=5,
    random_state=42
)

rf_model_3.fit(X_train, y_train)

rf_pred_3 = rf_model_3.predict(X_test)

print("R2:", r2_score(y_test, rf_pred_3))
```

```python
rf_model_4 = RandomForestRegressor(
    n_estimators=200,
    max_depth=10,
    random_state=42
)

rf_model_4.fit(X_train, y_train)

rf_pred_4 = rf_model_4.predict(X_test)

print("R2:", r2_score(y_test, rf_pred_4))
```

```python
rf_model_5 = RandomForestRegressor(
    n_estimators=200,
    min_samples_split=10,
    random_state=42
)

rf_model_5.fit(X_train, y_train)

rf_pred_5 = rf_model_5.predict(X_test)

print("R2:", r2_score(y_test, rf_pred_5))
```

```python
model_results = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Random Forest (100 trees)',
        'Random Forest (200 trees)',
        'Random Forest (Depth 5)',
        'Random Forest (Depth 10)',
        'Random Forest (Min Samples Split 10)'
    ],
    'R2 Score': [
        r2_score(y_test, y_pred),
        r2_score(y_test, rf_pred),
        r2_score(y_test, rf_pred_2),
        r2_score(y_test, rf_pred_3),
        r2_score(y_test, rf_pred_4),
        r2_score(y_test, rf_pred_5)
    ]
})

model_results = model_results.sort_values(
    by='R2 Score',
    ascending=False
)

print(model_results)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=model_results,
    x='R2 Score',
    y='Model'
)

plt.title('Model Performance Comparison')
plt.xlabel('R2 Score')
plt.ylabel('Model')
plt.show()
```

## Project conclusion

Developed a Student Performance Prediction project using the UCI Student Performance dataset. Performed exploratory data analysis, data preprocessing, categorical encoding, feature selection, and model training. Evaluated Linear Regression and Random Forest Regression using MAE, RMSE, and R². Random Forest with 200 estimators achieved the best test R² score of approximately 0.315 among the configurations tested. The results indicate that predicting final grades using demographic, family, and lifestyle features alone remains challenging.
