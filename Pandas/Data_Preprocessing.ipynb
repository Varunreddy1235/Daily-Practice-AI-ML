SimpleImputer (Scikit-learn)
In real Machine Learning projects, we often use SimpleImputer to fill missing values automatically.

Your next task: Import SimpleImputer from sklearn.impute.

```python
from sklearn.impute import SimpleImputer
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

# Create a sample DataFrame
df = pd.DataFrame({
    "Age": [21, np.nan, 22, 23, np.nan],
    "Marks": [80, 75, np.nan, 90, 85]
})

print("Original DataFrame:")
print(df)

# Create an imputer using the mean strategy
imputer = SimpleImputer(strategy="mean")

# Replace missing values
df[["Age", "Marks"]] = imputer.fit_transform(df[["Age", "Marks"]])

print("\nDataFrame after filling missing values:")
print(df)

# Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

# Create a sample DataFrame
df = pd.DataFrame({
    "Age": [21, np.nan, 22, 23, np.nan],
    "Marks": [80, 75, np.nan, 90, 85]
})

print("Original DataFrame:")
print(df)

# Create an imputer using the mean strategy
imputer = SimpleImputer(strategy="median")

# Replace missing values
df[["Age", "Marks"]] = imputer.fit_transform(df[["Age", "Marks"]])

print("\nDataFrame after filling missing values:")
print(df)

# Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

# Create a sample DataFrame
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", np.nan, "Sales", "IT", "HR"]
})
print("Original DataFrame:")
print(df)

# Create an imputer using the mean strategy
imputer = SimpleImputer(strategy="most_frequent")

# Replace missing values
df[['Department']] = imputer.fit_transform(df[["Department"]])

print("\nDataFrame after filling missing values:")
print(df)

# Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.DataFrame({
    "Department": ["IT", "HR", "Sales", "IT", "HR"]
})

Encoder_1 = OneHotEncoder(sparse_output=False)

encoded = Encoder_1.fit_transform(df[['Department']])

print(encoded)
print(Encoder_1.get_feature_names_out())
```

Your task:
- Create a OneHotEncoder.
- Encode the Department column.
- Get the encoded column names.
- Convert the encoded array into a DataFrame.
- Combine everything using pd.concat().
Expected final columns: Name, Department, Salary, Department_HR, Department_IT, Department_Sales.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.DataFrame({
    "Name": ["Ravi", "Raj", "Ram", "Varun", "Anu"],
    "Department": ["IT", "HR", "Sales", "IT", "HR"],
    "Salary": [30000, 25000, 35000, 40000, 28000]
})

Encoder = OneHotEncoder(sparse_output=False)

encoded_1 = Encoder.fit_transform(df[['Department']])

columns_names = Encoder.get_feature_names_out()

encoded_df = pd.DataFrame(
    encoded_1,
    columns=columns_names,
    index=df.index
)

Final_df = pd.concat([df, encoded_df], axis=1)

print(Final_df)
```
