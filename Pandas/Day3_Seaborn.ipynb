```python

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({
    "Student": ["Ravi", "Kumar", "Priya", "Anu", "Rahul"],
    "Marks": [85, 72, 95, 68, 88],
    "Hours": [4, 2, 6, 1, 5]
})
```

```python
#Bar plot — compare values
sns.barplot(data=df,x="Student",y="Marks")
```

```python
#scatter plot — relationship between two variables
sns.scatterplot(data=df,x="Hours",y="Marks")
```

```python
#Histogram — distribution of marks
sns.histplot(data=df,x="Marks",bins=5)
```

```python
#Box plot — spread of marks
sns.boxplot(data=df,x="Marks")
```

```python
#Heatmap — show values using colors
sns.heatmap(
    df[["Marks", "Hours"]].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.show()
```
