```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Python connected successfully!")
print("Matplotlib ready!")
```

```python
days = [1, 2, 3, 4, 5]
sales = [100, 200, 150, 300, 400]

plt.plot(days, sales)
plt.show()
```

```python
plt.plot(days,sales,marker='o',color='green',linestyle='--')
plt.title('Daily Sales Report')
plt.xlabel('days')
plt.ylabel('sales')
plt.show()
```

Lesson 3 — Bar Chart 📊
A bar chart helps compare different categories.

Example: Sales by product.

```python
products = ["Laptop", "Mobile", "Tablet"]
sales = [150000, 100000, 105000]
plt.bar(products,sales)
plt.title('Sales by Product Wise ')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

Practice Question 6
Which Matplotlib function creates a histogram?

```python
marks = [45, 55, 60, 65, 70, 75, 80, 85, 90]
plt.hist(marks,bins=5)
plt.title('Marks Distribution')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.grid()
plt.show()
```

```python
products = ["Laptop", "Mobile", "Tablet"]
sales = [50, 30, 20]
plt.pie(sales,labels=products,autopct='%1.1f%%')
plt.title('Product Sales')
plt.grid()
plt.show()
```

```python
plt.plot(products, sales)
plt.title("Daily Sales")
plt.grid()
plt.legend()
plt.show()
```

Exercise 1 — Customize a line chart
Use this data in your Jupyter notebook:
days = [1, 2, 3, 4, 5],
sales = [100, 150, 120, 200, 250]



Create a line chart with:
1. Title: Weekly Sales
2. X-axis label: Days
3. Y-axis label: Sales
4. Circle markers at each point
5. Grid lines

```python
days = [1, 2, 3, 4, 5]
sales = [100, 150, 120, 200, 250]
plt.plot(days,sales, marker="o")
plt.title('Weekly Sales')
plt.xlabel('Days')
plt.ylabel('Sales')
plt.grid()
plt.show()
```
