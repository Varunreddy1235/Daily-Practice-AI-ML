Mini project
Let's build a Student Performance Dashboard using Matplotlib. This will help you apply everything you've learned in one practical project.

We'll work one chart at a time.
Project data
students = ['Ravi', 'Kumar', 'Priya', 'Anu', 'Rahul']marks = [85, 72, 95, 68, 88]



Exercise 1 — Student marks bar chart
Create a vertical bar chart with:
- Title: Student Performance
- X-axis: Students
- Y-axis: Marks
- Different bar colors if you can.
- Display the chart.

```python
import matplotlib.pyplot as plt

students = ['Ravi', 'Kumar', 'Priya', 'Anu', 'Rahul']
marks = [85, 72, 95, 68, 88]
plt.bar(students,marks , color=['red', 'blue', 'green', 'orange', 'purple'])
plt.title('Student Performance ')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.grid(axis='y')
plt.show()
```

Exercise 2 — Find the highest-scoring student
Using the same students and marks lists:
1. Find the highest mark.
2. Find the student who scored it.
3. Print both results.
Hint: Try max() and marks.index()

```python
students = ['Ravi', 'Kumar', 'Priya', 'Anu', 'Rahul']
marks = [85, 72, 95, 68, 88]

# Find highest marks
highest = max(marks)

# Find the position of highest marks
position = marks.index(highest)

# Find the student's name
topper = students[position]

# Display results
print("Highest Marks:", highest)
print("Top Student:", topper)
```

Average marks? 

```python
average = sum(marks) / len(marks)
print("Average Marks:", average)
```

Exercise 3 — Find the lowest-scoring student
Use the same students and marks lists.

Your task is to print:
Lowest Marks: 68
Lowest Scoring Student: Anu


Hint: Use the same logic as the highest-scoring student, but replace max() with min().

```python
lowest_score =min(marks)
position=marks.index(lowest_score)
Student=students[position]
print('lowest_score:', lowest_score)
print('Student:',Student)
```

Exercise 4 — Find students who scored above 80
Expected output:
Ravi: 85
Priya: 95
Rahul: 88


Hint: Use a for loop and an if condition.

```python
for i in range(len(marks)):
    if marks[i]>80:
        print(students[i],':', marks[i])
```

Exercise 5 — Add a result to the dashboard
Now calculate the percentage of students who scored above 80.

For example:
- Total students = 5
- Students above 80 = 3
- Percentage = 60%
Hint: Use a counter inside your for loop, then calculate:
percentage=count/total students *100
Try writing the code yourself.

```python
count=0
for i in range(len(marks)):
    if marks[i]>80:
        count=count+1
percentage=count/len(students)*100
print("Students above 80:", count)
print("Percentage:", percentage, "%")
```
